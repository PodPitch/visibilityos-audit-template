# Reproduce the VisibilityOS AI answer study totals

[VisibilityOS by PodPitch](https://visibilityos.co/) published a retrospective study of 817 matched comparisons across eight AI engine views. This notebook uses the study's published aggregate counts to reproduce the per-view percentages and the pooled result.

A comparison counts as a change if the tracked brand was mentioned or linked on just one of the two dates. Read the [full study and matching method](https://visibilityos.co/research/ai-answer-visibility) for the settings held constant, date coverage, and limitations.


```python
# Engine view, matched comparisons, changes in brand presence.
engine_views = [
    ("ChatGPT", 107, 6),
    ("ChatGPT web", 107, 3),
    ("Claude", 107, 14),
    ("Gemini", 107, 9),
    ("Gemini web", 106, 15),
    ("Google AI Mode", 95, 7),
    ("Grok", 81, 21),
    ("Perplexity", 107, 7),
]

print(f'{"Engine view":16} {"Checks":>7} {"Changes":>8} {"Share changed":>14}')
for view, checks, changes in sorted(engine_views, key=lambda row: row[2] / row[1], reverse=True):
    print(f"{view:16} {checks:7d} {changes:8d} {changes/checks:14.1%}")

total_checks = sum(checks for _, checks, _ in engine_views)
total_changes = sum(changes for _, _, changes in engine_views)
assert (total_checks, total_changes) == (817, 82)
print(f"{'Pooled result':16} {total_checks:7d} {total_changes:8d} {total_changes/total_checks:14.1%}")
```

## What the totals mean

The pooled rate divides all 82 changes by all 817 matched comparisons. Engine views with more comparisons contribute more to that rate. Some checks contribute to more than one comparison, and the views have different date coverage. The results describe this sample of PodPitch-owned sites. They cannot isolate an engine effect or establish which engine is more reliable overall.
